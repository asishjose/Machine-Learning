```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('daily-min-temp.csv', usecols=['min_temp'])
data.head()
```

```python
data['min_temp'] = pd.to_numeric(data['min_temp'], errors='coerce')
```

```python
def create_sequence(data, seq_length):
    X,y = [],[]
    for i in range(len(data)):
        X.append(data[i:seq_length+1])
        y.append(data[seq_length])
    return np.ndarray(X), np.ndarray(y)

seq_length = 12

```

```python
X,y = create_sequence(data, seq_length)
```
